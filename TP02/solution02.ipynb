#  Problem 2. My Shopping List

##### A list can grow and shrink. Build a shopping list and change it step by step. Print the list after each step to see what happened.

```python
# Step 1. Create the list with three items and print it. 
shopping= ["rice", "eggs", "milk"]
print("Start : ", shopping)
```

```python
#Step 2. Add "bread" at the end with append(), and "water" at the front (position 0) with insert().
shopping.append("bread")
print(shopping)
shopping.insert(0,"water")
print("Added : ", shopping)
```

```python
#Step 3. Remove "eggs" by its value with remove(). Then remove the last item with pop(), which also gives it back to you.
shopping.remove("eggs")
last = shopping.pop()
print("Removed : ", last)
print("Now  :  ", shopping)
```

```python
# Step 4. Use slicing list[start:stop:step] to print the first 2 items, the last 2 items and the reversed list.
print("First 2 : ", shopping[:2])
print("Last 2 : ", shopping[-2:])
print("Reversed : ", shopping[::-1])
```

```python
# Step 5. Check whether "milk" is in the list with in, then print each item with a number using enumerate().
print("Is milk in the list ? ", "milk" in shopping)
for i, item in enumerate(shopping, start=1):
    print(i, item)
```

##### Answer for the group discussion

1.What is the difference between remove() and pop()?

Answer : The differeces are :
- remove(): We use it when we want to remove a specific item from a list by its value. It removes the first matching item and does not return the removed item (it returns None).
- pop(): We use it to remove an item from a list by its index. By default, it removes the last item and returns the removed item. We can also specify an index.

```python
try:
    shopping.remove("fish")
except ValueError as error:
    print("Error : ", error)
```

2.What happens if you call shopping.remove("fish")? Try it and read the error. 

Answer : 
- If we use shopping.remove("fish"), Python will raise a ValueError because "fish" does not exist in the list.
- This happens because the remove() method can only remove an item that exists in the list.

3. Does slicing (shopping[:2]) change the original list? How can you check?

Answer : 
- No, slicing does not change the original list. It creates and returns a new list containing the selected items.
- We can check this by printing the original list before and after slicing.
