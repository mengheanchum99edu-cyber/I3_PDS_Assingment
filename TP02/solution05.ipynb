#  Problem 5. Fruit Shop

#### A fruit shop keeps its prices in a dictionary: the key is the fruit, the value is the price in dollars.

```python
# Step 1. Create the dictionary and read the price of a mango with its key.
prices = {"apple": 0.50, "banana": 0.25, "mango": 1.00}
print("Price of mango : ", prices["mango"])
```

```python
# Step 2. Add oranges at $0.75, change the apple price to $0.60, and delete bananas with del.
prices["orange"] = 0.75
prices["apple"] = 0.60
del prices["banana"]
print("Price : ", prices)
```

```python
#Step 3. prices["grape"] would crash because the key does not exist. get() returns a default value instead.
print("Price of gape : ", prices.get("gape", "note sold"))
```

```python
#Step 4. Print a price list with a loop over items(), which gives each (key, value) pair.
for fruit, price in prices.items():
    print(f"{fruit:<7} ${price:0.2f}")
```

```python
# Step 5. A customer buys 4 apples, 2 mangoes and 3 oranges. Compute the bill: for each fruit, add price × quantity to total.
basket = {"apple": 4, "mango": 2, "orange": 3}
total = 0 
for fruit, qty in basket.items():
    total = total + prices[fruit] * qty
print(f"Total bill : ${total:0.2f}")
```

##### Answer of Discussion

1.1. Why is a dictionary better than two lists (fruits and prices) for this shop?

Answer : 
- A dictionary is better because it stores fruit names and their prices together as key-value pairs.
- We can easily find a fruit's price by using its name as a key, such as prices["apple"].

If we use two separate lists, we need to remember the position (index) of each fruit and its corresponding price. This can cause mistakes if the lists are not in the same order.

Therefore, dictionaries make our code easier to read, manage, and update.

2. What is the difference between prices["grape"] and prices.get("grape")? Try both.

Answer : Both are used to access values in a dictionary, but they behave differently when the key does not exist.
- prices["grape"]: If the key "grape" does not exist, Python raises a KeyError, which stops the program unless the error is handled.
- prices.get("grape"): If the key does not exist, it returns None instead of causing an error.
