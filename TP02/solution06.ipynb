# Group 6  ·  Problem 6. Mini Grade Book

##### This problem uses all four structures. Each student (key) has a list of three scores (value). You will compute averages, collect the students who passed in a set, and rank everyone with tuples.

```python
# Step 1. Create the grade book: a dictionary whose values are lists.
grades = {
    "Dara" : [80, 75, 90],
    "Sophea" : [95, 88, 92],
    "Vibol" : [40, 55, 45],
    "Malis" : [70, 85, 78]
}
```

```python
# Step 2. Compute each student's average and store it in a new dictionary averages.
averages = {}
for name, scores in grades.items():
    averages[name] = round(sum(scores) / len(scores), 1)
print("Averages : ", averages)
```

```python
# Step 3. The pass mark is 50. Print PASS or FAIL for each student and put the students who passed in a set. 
passed = set()

for name, avg in averages.items():
    if avg >= 50 :
        print(name , "PASS")
        passed.add(name)
    else:
        print(name, "FAIL")
print("Passed : ", sorted(passed))
```

```python
# Step 4. Rank the students. averages.items() gives (name, average) tuples; sort them by the average (index 1), best first, and number them with enumerate().
ranking = sorted(averages.items(), key = lambda item: item[1], reverse = True)
for rank, (name , avg) in enumerate(ranking, start = 1):
    print(f"#{rank} {name} ({avg})")
```

#### Answer of Discussion

1. Name the four data structures used in this problem and say why each one was chosen.

Answer : There are four main data structures used in this problem:
- Dictionary (dict) : We use a dictionary to store each student's name as a key and their scores as a value. It makes it easy to find and manage each student's grades.
- List (list) : We use a list to store three scores for each student. A list allows us to keep multiple score together and calculate their average.
- Set (set) : We use a set to store the names of students who passed. A set keeps only unique value, so a student cannot appear more than once.
- Tuple (tuple) : We use tuples when ranking students because averages.items() returns pairs of values (student name and average). We can sort these pairs by average score.

Each data structure has a different purpose, which makes the program easier to organize and manage.
