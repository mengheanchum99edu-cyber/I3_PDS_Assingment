```python
# Step01: Create two lists: the days and their temperatures. Position 0 of days matches position 0 of temps. 
days =["Mon", "Tue","Wed","Thu", "Fri", "Sat", "Sun"]
temperature = [31, 33, 29, 34, 35, 32, 30]
```

```python
#Step02: Print the number of days with len(). Print the first day and the last day. Remember: the first index is 0 and the last index is -1. 
print("Number of days: ", len(days))
print("First day: ", days[0], temperature[0])
print("Last day : ", days[6], temperature[6]) # we also can use days[-1]
print("Last day : ", days[-1], temperature[-1])
```

```python
#Step03: Print the highest, the lowest and the average temperature. The average is the sum divided by the number of items.
print("Highest: ", max(temperature))
print("Lowest : ", min(temperature))
average = sum(temperature) / len(temperature)
print("Average : ", round(average, 1))
```

```python
#step04:  Find the day of the highest temperature. temps.index(x) gives the position of x; use that position in days.
highest = temperature.index(max(temperature))
print(highest)
print("Hottest day : ", days[highest],temperature[highest] ) # day[4]
```

```python
#step05: . Print the temperatures from lowest to highest with sorted(), then print temps again to show the original list did not change. 
print("Sorted : ",sorted(temperature))
print("Original : ", temperature)
```

### Answer to the question

##### 1.Why do we use two lists here, and why must they be in the same order?

Answer: The reason are :
- We use two lists to store different types of information: days stores the days of the week, and temps stores the temperatures.
- They must be in the same order because each temperature must match its corresponding day at the same index.

##### 2. What is the difference between sorted(temps) and temps.sort()?

- sorted(temps) returns a new sorted list without changing the original list.
- temps.sort() sorts the original list directly and changes its order. It does not return a new sorted list (it returns None).
