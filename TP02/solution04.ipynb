# Problem 04: Problem 4. Two Clubs

##### The school has a football club and a music club. Some students joined both. A set keeps each name only once, which makes it easy to compare the two clubs. 

```python
#Step 1. The football list has a duplicate ("Dara" registered twice). Turn it into a set and compare the sizes. 
football_list = ["Dara", "Vibol", "Kosal", "Dara", "Bopha"]
football = set(football_list)
print("List size : ", len(football_list), "| Set size : ", len(football))
```

```python
#Step 2. Create the music club directly as a set with curly braces. 
music = {"Sophea", "Kosal", "Malis", "Dara"}
```

```python
#Step 3. Use the four set operators. A set has no order, so wrap each result in sorted() to print it alphabetically.
print("Both club         : ", sorted(football & music)) # intersection
print("Any Club          : ", sorted(football | music)) # union
print("Footbal Only      : ", sorted(football - music)) # difference
print("Exactly one       : ", sorted(football ^ music)) # symmetric difference
```

```python
#Step 4. Add "Chenda" to the music club and remove "Malis". Then check whether "Vibol" is in the music club.
music.add("Chenda")
music.discard("Malis")
print("Music now      : ", sorted(music))
print("Is Vibol in music ? ", "Vibol" in music)
# Deletes a specific element. If the element does not exist, Python raises a KeyError.
# Deletes a specific element. If the element does not exist, Python does not raise an error.
```

#### Answer of the discussion

1. Why did the set have 4 names and not 5?

Answer : 
- Because a set does not allow duplicate elements.
- If we add the same name more than once, Python keeps only one copy.
- Therefore, even if we add 5 names, the set may contain only 4 unique names.

2. Print football without sorted() several times. Is the order always the same? Why can we not write football[0]?

Answer : 
- No, we should not expect the order of a set to be the same every time because sets are unordered collections.
- We cannot write football[0] because sets do not support indexing.
- If we want to display the elements in order, we can use sorted().
